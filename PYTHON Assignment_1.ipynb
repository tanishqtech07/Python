# Part A – Theory Questions

Q1. What is a variable in Python?

```python
# Write your Answer with Example
"""the variable in which we store the value of different data types and use it in program, we can also take user input and store it in a variable and use it in our program. for example:"""
a=10#a is a variable which stores the value 10
print(a)
b="hello"#b is a variable which stores the value "hello"
print(b)
```

Q2. What are Data Types in Python? Name all Commonly used data types in Python?

```python
# Write your Answer
"""in python we can use these data types to store different types of data in our program.the common data types in python are int,float,str,list,tuple,set,dictionary. for example:"""
a=10#int data type which stores whole numbers
b=3.14#float data type which stores decimal numbers
c="hello"#str data type which stores string of characters
print(type(a))
print(type(b))
print(type(c))
```

Q3. What is the difference between int and float?

```python
'''in python int is data type store the integer values not decimal values and the float data type store the decimal values or fractional values ,
when division of two integers is done then the result is always a float value. for example:'''
print(10/2)#the result is 5.0 which is a float value
print(7/3)#the result is 2.3333333333333335 which is a float value
```

Q4. What is type() function?

```python
"""the type() function to check the data type o a variable that which type of data is stored in a variable. for example:"""
a=10
print(type(a))#the result is <class 'int'> which means the data type of variable a is int
```
